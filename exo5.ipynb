# Challenger O-ring analysis

The goal is to find a problem in the first analysis.

## 1. Load the data

The data contains past O-ring tests. Each row gives the temperature, pressure and the number of failures.

```python
import pandas as pd

file_path = '../../mooc-rr-ressources/cruft/module5/ressources/moocrr_notebook/shuttle.csv'
data = pd.read_csv(file_path)
data
```

## 2. Check tests

Count all tests and the tests with no failure.

```python
print('Number of tests:', len(data))
print('Tests with no failure:', (data['Malfunction'] == 0).sum())
print('Total O-rings:', data['Count'].sum())
print('Total failures:', data['Malfunction'].sum())
```

## 3. Temperature comparison

The launch temperature was expected to be 31°F. 
A lower temperature is important because the O-rings may behave differently in colder conditions.

```python
launch_temperature = 31
lowest_test_temperature = data['Temperature'].min()

print('Launch temperature:', launch_temperature, '°F')
print('Lowest test temperature:', lowest_test_temperature, '°F')
```

## 4. Possible problem

The first analysis removed all rows with zero failures. A test with zero failures is useful information.
If we remove these rows, the risk can look smaller than it is.

```python
selected_data = data[data['Malfunction'] > 0]

print('Rows in the full data:', len(data))
print('Rows after removing zero failures:', len(selected_data))
```

## 5. Keeping all rows

The overall failure frequency is the total number of failures divided by the total number of O-rings.

Comparing the tests with failures and tests with zero failures, we can see how much the result changes

```python
overall_frequency = data['Malfunction'].sum() / data['Count'].sum()
selected_frequency = selected_data['Malfunction'].sum() / selected_data['Count'].sum()

print('Frequency using all tests:', overall_frequency)
print('Frequency after removing zero-failure tests:', selected_frequency)
```

## Conclusion

The first analysis removed all tests with zero failures. This is a problem because these tests also give information about the risk.

After removing them, the data no longer represent all the tests. They represent only the tests where at least one failure was already seen. This can make the estimated risk misleading.
